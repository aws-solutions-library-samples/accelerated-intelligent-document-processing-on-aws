# Step 3: Information Extraction Using Agentic Methods


This notebook performs information extraction from classified document sections using AWS Bedrock.

This extraction method utilises Strands Agent for a self correcting mechanism working against pydantic models internally to ensure schema adherence.

**Important**

This method only works with higher intelligence models that support tool use. Recommended model family is Anthropic Claude.
OpenAI models should perform well too but they are limited to text input only inside bedrock.
For Amazon models you should use Nova Pro.

**Inputs:**
- Document object with classification results from Step 2
- Extraction configuration
- Document classes with attributes definition

**Outputs:**
- Document with extraction results for each section
- Structured data extracted based on document class attributes

This notebook also demonstrates **schema-constraint validation** (full
JSON-Schema checks including `format` keywords) and optional **model escalation**
for results that fail validation — see the *Schema-Constraint Validation & Model
Escalation* section below.

### 3. Extract Information from Document Sections

```python
# Helper function to parse S3 URIs and load JSON
def parse_s3_uri(uri):
    parts = uri.replace("s3://", "").split("/")
    bucket = parts[0]
    key = "/".join(parts[1:])
    return bucket, key

def load_json_from_s3(uri):
    s3_client = boto3.client('s3')
    bucket, key = parse_s3_uri(uri)
    response = s3_client.get_object(Bucket=bucket, Key=key)
    content = response['Body'].read().decode('utf-8')
    return json.loads(content)

print("Helper functions defined")
```

```python
print("Extracting information from document sections...")

if not document.sections:
    print("No sections found in document. Cannot proceed with extraction.")
else:
    extraction_results = []
    
    # Process each section (limit to first 3 to save time in demo)
    n = min(3, len(document.sections))
    print(f"Processing first {n} of {len(document.sections)} sections...")
    
    for i, section in enumerate(document.sections[:n]):
        print(f"\n--- Processing Section {i+1}/{n} ---")
        print(f"Section ID: {section.section_id}")
        print(f"Classification: {section.classification}")
        print(f"Pages: {section.page_ids}")
        
        # Process section extraction
        start_time = time.time()
        document = extraction_service.process_document_section(
            document=document,
            section_id=section.section_id
        )
        extraction_time = time.time() - start_time
        
        print(f"Extraction completed in {extraction_time:.2f} seconds")
        
        # Record results
        extraction_results.append({
            'section_id': section.section_id,
            'classification': section.classification,
            'processing_time': extraction_time,
            'extraction_result_uri': getattr(section, 'extraction_result_uri', None)
        })
    
    print(f"\nExtraction complete for {n} sections.")
```

```python
print("\n=== Extraction Results ===")

if document.sections:
    for i, section in enumerate(document.sections[:n]):
        print(f"\n--- Section {section.section_id} ({section.classification}) ---")
        
        if hasattr(section, 'extraction_result_uri') and section.extraction_result_uri:
            try:
                # Load extraction results from S3
                extraction_data = load_json_from_s3(section.extraction_result_uri)
                
                print(f"Extraction Result URI: {section.extraction_result_uri}")
                
                # Display inference results
                if 'inference_result' in extraction_data:
                    inference_result = extraction_data['inference_result']
                    print("Extracted Data:")
                    for attr_name, attr_value in inference_result.items():
                        if attr_value is not None:
                            # Truncate long values for display
                            display_value = str(attr_value)[:1000] + "..." if len(str(attr_value)) > 1000 else attr_value
                            print(f"  {attr_name}: {display_value}")
                        else:
                            print(f"  {attr_name}: null")
                else:
                    print("No inference results found")
                    
                # Display metadata if available
                if 'metadata' in extraction_data:
                    metadata = extraction_data['metadata']
                    print(f"Processing time: {metadata.get('extraction_time_seconds', 'N/A')} seconds")
                    
            except Exception as e:
                print(f"Error loading extraction results: {e}")
        else:
            print("No extraction results available")
else:
    print("No sections to display")
```

## Schema-Constraint Validation & Model Escalation

Agentic extraction always validates the agent's output against the **Pydantic
model** generated from the class schema (so `enum`, `pattern`, numeric bounds and
`minItems` are enforced and self-corrected during extraction). Those constraints
are enforced at the agent's `extraction_tool` boundary rather than merely
reported, so a `minItems` floor the agent cannot reach fails the section with no
rows kept — see the caveat in part C. The `extraction.validation` block adds two
things on top:

1. **Full JSON-Schema validation of the result** — notably the `format` keyword
   (`date`, `email`, `uuid`, ...), which the Pydantic model does *not* enforce.
   All violations are reported in one pass with field paths.
2. **Bounded model escalation** — when validation still fails, optionally
   re-extract **once** with a stronger model (cheaper than human review). This is
   opt-in: it runs only under `fail_action: escalate`.

Config keys live under `extraction.validation`, a sibling of `extraction.agentic`
rather than a key inside it, because Simple extraction runs the same
validate-and-retry path: `enabled` (default `true`), `check_formats` (default
`true`), `fail_action` (`warn` | `escalate` | `reject`, default `warn`),
`escalation_model` (default none; per-class override:
`x-aws-idp-extraction-escalation-model`), and `min_population_ratio` (default
`0.5`, an advisory completeness warning). A stored configuration written against
the older `extraction.agentic.validation` path is relocated on read, so nothing
breaks on upgrade; author new configuration at `extraction.validation`.

⚠️ **A misspelled or mis-nested key under `extraction` is accepted in
silence.** Unknown keys are ignored by the config models, and the unknown-key
warning covers the top level of the configuration only, so
`validation.enabld: false` or a typo in the block name is dropped with no log line
and the setting you think you changed is simply not set. Only a bad *value* on a
known field fails loudly. Read the setting back off the loaded config, as part B
does, rather than off the dict you wrote.

> A dedicated, AWS-free deep-dive lives in
> `notebooks/misc/agentic-extraction-validation-and-escalation.ipynb`.

### B. Enable validation + escalation on the agentic service

Validation is on by default; we set the block explicitly here so the config is
self-describing, then re-run extraction on the first section. The validator runs
**inside** the agent loop (feeding any violations back for self-correction) and
again on the final result; the outcome is recorded under `metadata.validation` in
the section's extraction result.

`fail_action` defaults to `warn`, which records the outcome and costs nothing. Set
`fail_action: "escalate"` and an `escalation_model` to opt into re-extracting
failing results with a stronger model. Here we ask for `escalate` and leave
`escalation_model` unset so it falls back to the extraction model (a plain second
attempt) — set it to e.g. `us.anthropic.claude-opus-4-8` in your own runs.

The last line of the next cell reads the block back off the service's own loaded
config rather than off `CONFIG_VALIDATED`, so a mis-nested or misspelled key shows
up as an unchanged default instead of being echoed back at you.

```python
from copy import deepcopy

# Enable validation on a copy of the config. `validation` is a sibling of
# `agentic` under `extraction`, not a key inside it.
CONFIG_VALIDATED = deepcopy(CONFIG)
CONFIG_VALIDATED["extraction"]["agentic"] = {"enabled": True}
CONFIG_VALIDATED["extraction"]["validation"] = {
    "enabled": True,
    "check_formats": True,      # enforce date/email/uuid formats
    "fail_action": "escalate",  # warn (the default, and free) | escalate | reject
    # "escalation_model": "us.anthropic.claude-opus-4-8",  # stronger tier
}
CONFIG_VALIDATED["extraction"]["model"] = "us.anthropic.claude-sonnet-4-6"

extraction_service_validated = extraction.ExtractionService(config=CONFIG_VALIDATED)
print("Validation-enabled extraction service initialized")
# Read the loaded config, not the dict above: an unknown or mis-nested key is
# ignored silently, so only the service's own view shows what is in effect.
print("validation config:", extraction_service_validated.config.extraction.validation)
```

```python
# Re-extract the first section with validation enabled.
if document.sections:
    test_section = document.sections[0]
    print(f"Extracting (validated) section {test_section.section_id} "
          f"[{test_section.classification}]...")

    doc_validated = extraction_service_validated.process_document_section(
        document=deepcopy(document),
        section_id=test_section.section_id,
    )

    vsection = doc_validated.sections[0]
    if getattr(vsection, "extraction_result_uri", None):
        vdata = load_json_from_s3(vsection.extraction_result_uri)
        vmeta = vdata.get("metadata", {})
        validation_meta = vmeta.get("validation")

        print("\n=== Validation outcome (metadata.validation) ===")
        if validation_meta is None:
            print("No 'validation' block — result passed all schema constraints, "
                  "or validation was disabled.")
        else:
            print(json.dumps(validation_meta, indent=2))
            print(f"\nvalid            : {validation_meta.get('valid')}")
            print(f"escalated        : {validation_meta.get('escalated')}")
            print(f"escalation_model : {validation_meta.get('escalation_model', '(none)')}")
            print(f"failed_fields    : {validation_meta.get('failed_fields')}")
    else:
        print("No extraction result URI produced.")
else:
    print("No sections available to extract.")
```

### C. `fail_action` cheat-sheet

| `fail_action` | When the result is still invalid after self-correction |
|---------------|---------------------------------------------------------|
| `warn` | the default: record a ProcessingIssue and `metadata.validation`, then proceed — free, no extra inference |
| `escalate` | re-extract once with `escalation_model`; keep it only if valid / fewer errors, then warn. The one action that spends another Bedrock call |
| `reject` | as `warn`, plus record `parsing_succeeded = false` in the section result |

> ⚠️ **None of the three rejects a section or routes it to human
> review.** `parsing_succeeded` is read by the extraction processing report, which
> prints `FAILED` for the section, and by the Processing Report tab in the web UI
> — and by nothing in the status path. So under `reject` the extracted values are
> still written to `result.json`, the section completes, the document still reports
> `COMPLETED`, and no review task is created. Human review is routed by the
> top-level `hitl` block (`hitl.enabled` plus per-field confidence against
> `hitl.confidence_threshold`), which never looks at `parsing_succeeded`. The one
> extraction setting that turns a detection into an outcome is
> `extraction.row_shortfall_action: fail`, which fails the Step Functions
> execution after persisting the rows, the issue and the report.

> **`format: date` caveat:** JSON-Schema `format: date` means ISO-8601
> (`YYYY-MM-DD`). The default extraction prompt asks for `MM/DD/YYYY`, which will
> fail format validation. Set `check_formats: false` or use a `pattern` for
> non-ISO dates. See the extraction README for full details.

> ⚠️ **`minItems` caveat:** on this agentic path `minItems` is a hard
> floor, not a warning. It survives into the Pydantic model the agent's
> `extraction_tool` validates against, so a short list is rejected inside the tool,
> the agent spends its correction rounds on it, and if it still cannot reach the
> floor the section fails with **no rows kept**. Pick a floor you are willing to
> fail the document on, and put none on the lists of a section that shards: with
> `extraction.agentic.max_concurrent_batches > 1` (the shipped configuration sets
> it to 10, with `max_pages_per_shard: 5`) each shard agent's tool is built from
> the shard transport model, which relaxes *presence* for a required container but
> carries every row-count bound unchanged,
> while seeing only its own page range, so a section-sized floor such as
> `minItems: 100` rejects every shard that holds fewer rows — and a cover-page
> shard holds none. For a completeness signal that keeps the rows, use
> `extraction.row_shortfall_action` (`warn` | `fail`, default `warn`), which is
> shard-aware. No shipped preset sets `minItems`.

```python
# Let's extract from the same section using both methods for direct comparison
from copy import deepcopy


if document.sections and len(document.sections) > 0:
    test_section = document.sections[0]  # Use first section
    
    print(f"📋 Testing with Section: {test_section.classification}")
    print("="*60)
    
    # Track metrics
    comparison_results = {}
    
    # Method 1: Traditional Extraction
    print("\n🔴 METHOD 1: TRADITIONAL EXTRACTION")
    print("-"*40)
    
    CONFIG_TRAD = CONFIG.copy()
    CONFIG_TRAD["extraction"]["agentic"] = {"enabled": False}
    
    extraction_service_traditional = extraction.ExtractionService(config=CONFIG_TRAD)
    
    start_time = time.time()
    try:
        document_trad = extraction_service_traditional.process_document_section(
            document=deepcopy(document),
            section_id=test_section.section_id
        )
        trad_time = time.time() - start_time
        
        # Load results
        trad_section = document_trad.sections[0]
        if trad_section.extraction_result_uri:
            trad_data = load_json_from_s3(trad_section.extraction_result_uri)
            trad_result = trad_data.get('inference_result', {})
            
            comparison_results['traditional'] = {
                'time': trad_time,
                'fields_extracted': len([k for k, v in trad_result.items() if v is not None]),
                'total_fields': len(trad_result),
                'has_nested': any(isinstance(v, dict) for v in trad_result.values()),
                'has_arrays': any(isinstance(v, list) for v in trad_result.values())
            }
            
            print(f"✅ Completed in {trad_time:.2f} seconds")
            print(f"   Fields: {comparison_results['traditional']['fields_extracted']}/{comparison_results['traditional']['total_fields']}")
            print(f"   Complex structures: Nested={comparison_results['traditional']['has_nested']}, Arrays={comparison_results['traditional']['has_arrays']}")
    except Exception as e:
        print(f"❌ Failed: {e}")
        comparison_results['traditional'] = {'error': str(e)}
    
    # Method 2: Agentic Extraction
    print("\n🟢 METHOD 2: AGENTIC EXTRACTION")
    print("-"*40)
    
    CONFIG_AGENT = CONFIG.copy()
    CONFIG_AGENT["extraction"]["agentic"] = {"enabled": True}
    
    extraction_service_agentic = extraction.ExtractionService(config=CONFIG_AGENT)
    
    start_time = time.time()
    try:
        document_agent = extraction_service_agentic.process_document_section(
            document=deepcopy(document),
            section_id=test_section.section_id
        )
        agent_time = time.time() - start_time
        
        # Load results
        agent_section = document_agent.sections[0]
        if agent_section.extraction_result_uri:
            agent_data = load_json_from_s3(agent_section.extraction_result_uri)
            agent_result = agent_data.get('inference_result', {})
            
            comparison_results['agentic'] = {
                'time': agent_time,
                'fields_extracted': len([k for k, v in agent_result.items() if v is not None]),
                'total_fields': len(agent_result),
                'has_nested': any(isinstance(v, dict) for v in agent_result.values()),
                'has_arrays': any(isinstance(v, list) for v in agent_result.values())
            }
            
            print(f"✅ Completed in {agent_time:.2f} seconds")
            print(f"   Fields: {comparison_results['agentic']['fields_extracted']}/{comparison_results['agentic']['total_fields']}")
            print(f"   Complex structures: Nested={comparison_results['agentic']['has_nested']}, Arrays={comparison_results['agentic']['has_arrays']}")
    except Exception as e:
        print(f"❌ Failed: {e}")
        comparison_results['agentic'] = {'error': str(e)}
    
    # Comparison Summary
    print("\n📊 COMPARISON SUMMARY")
    print("="*60)
    
    if 'traditional' in comparison_results and 'agentic' in comparison_results:
        if 'error' not in comparison_results['traditional'] and 'error' not in comparison_results['agentic']:
            trad = comparison_results['traditional']
            agent = comparison_results['agentic']
            
            speed_diff = ((trad['time'] - agent['time']) / trad['time']) * 100
            field_diff = agent['fields_extracted'] - trad['fields_extracted']
            
            print(f"⏱️  Speed: Agentic is {speed_diff:.1f}% {'faster' if speed_diff > 0 else 'slower'}")
            print(f"    Traditional: {trad['time']:.2f}s | Agentic: {agent['time']:.2f}s")
            print()
            print(f"📝 Field Extraction: Agentic extracted {field_diff:+d} more fields")
            print(f"    Traditional: {trad['fields_extracted']}/{trad['total_fields']} | Agentic: {agent['fields_extracted']}/{agent['total_fields']}")
            print()
            print(f"🏗️  Structure Handling:")
            print(f"    Nested Objects: Traditional={trad['has_nested']} | Agentic={agent['has_nested']}")
            print(f"    Arrays: Traditional={trad['has_arrays']} | Agentic={agent['has_arrays']}")
            
            print("\n✨ KEY ADVANTAGES OF AGENTIC EXTRACTION:")
            advantages = []
            if speed_diff > 10:
                advantages.append(f"• {speed_diff:.0f}% faster processing")
            if field_diff > 0:
                advantages.append(f"• Better field coverage (+{field_diff} fields)")
            if agent['has_nested'] or agent['has_arrays']:
                advantages.append("• Handles complex nested structures")
            advantages.append("• Self-correcting with schema validation")
            advantages.append("• Consistent output format guaranteed")
            
            for adv in advantages:
                print(adv)
    else:
        print("⚠️ Comparison could not be completed due to errors")
```
